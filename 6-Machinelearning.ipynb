## Q1. Image Classification Basics (5 marks)

**Q:** What is the main challenge in image classification, and why can’t we hard-code rules to classify images?

**A:**  
- **Challenge:** The **semantic gap** — images are just pixel grids (e.g., $1280 \times 720 \times 3$), but classes like *cat* or *car* are high-level semantic concepts.  
- **Hard-coding fails** due to variations in:  
  - Illumination  
  - Scale  
  - Viewpoint  
  - Occlusion  
  - Background clutter  
- Instead, we use **data-driven classifiers** trained on labeled datasets.

## Q2. Nearest Neighbor Classifier (10 marks)

**Q:** Describe the training and testing complexity of the 1-NN classifier.

**A:**  
- **Training:** Just memorize data → Complexity $O(1)$ (constant time).  
- **Testing:** For each test sample, compute distance to all $N$ training samples → Complexity $O(N)$ (linear time).  
- **Issue:** Testing must be fast, but 1-NN is slow.

## Q3. Distance Metrics (5 marks)

**Q:** Write equations for L1 and L2 distance between two vectors $x,y \in \mathbb{R}^d$.

**A:**  

- **L1 (Manhattan):**  
  $$
  d_{L1}(x,y) = \sum_{i=1}^{d} |x_i - y_i|
  $$

- **L2 (Euclidean):**  
  $$
  d_{L2}(x,y) = \sqrt{\sum_{i=1}^{d} (x_i - y_i)^2}
  $$

- **Effect:** Choice of metric affects decision boundaries.

## Q4. K-Nearest Neighbors (10 marks)

**Q:** Why does increasing $K$ smooth decision boundaries?

**A:**  
- $K=1$: Boundary is highly irregular, sensitive to outliers.  
- $K>1$: Majority voting reduces noise effect → smoother regions.  
- Example: For $K=3$, test sample classified by majority of 3 nearest neighbors.  
- **Risk:** Large $K$ → oversmoothing → underfitting.

## Q5. Curse of Dimensionality (10 marks)

**Q:** Explain the curse of dimensionality and why KNN on raw pixels is rarely used.

**A:**  
- In high dimensions, distances become less meaningful.  
- Required number of samples grows **exponentially** with dimension.  
- Example: $32 \times 32$ image → 1024 dimensions per channel → impossible coverage.  
- Raw pixel distances uninformative: visually different images can have same $L2$ distance.

## Q6. Linear Classifier (10 marks)

**Q:** Define the linear classifier score function for classification.

**A:**  
For input $x \in \mathbb{R}^d$, weights $W \in \mathbb{R}^{C \times d}$, bias $b \in \mathbb{R}^C$:  

$$
f(x; W, b) = Wx + b
$$

- Output: class scores for $C$ classes.  
- Prediction:  
  $$
  \hat{y} = \arg\max_j f_j(x)
  $$

## Q7. Multiclass SVM Loss (15 marks)

**Q:** Write the SVM loss function and explain.

**A:**  
For training sample $(x_i, y_i)$:  

$$
L_i = \sum_{j \neq y_i} \max\big(0, f_j(x_i) - f_{y_i}(x_i) + \Delta\big)
$$

- Enforces margin $\Delta$ between correct class score and incorrect ones.  
- If margin satisfied → loss = 0.  
- If violated → penalty increases.  

---

## Q8. Cross-Entropy Loss (10 marks)

**Q:** Derive the cross-entropy loss for softmax probabilities.

**A:**  

- **Softmax probability for class $j$:**

$$
p_j = \frac{e^{f_j(x)}}{\sum_k e^{f_k(x)}}
$$

- **Cross-entropy loss:**

$$
L_i = -\log p_{y_i}
$$

- Encourages high probability for true class.  
- Properties:  
  - Minimum loss = 0 (perfect prediction).  
  - If uniform probabilities:  
    $$
    L_i = \log(C)
    $$
    Example: $C=10 \implies L_i \approx 2.3$.

---

## Q9. Regularization (10 marks)

**Q:** Why do we use regularization in linear classifiers?

**A:**  
- Without regularization: Many $W$ may achieve zero training error.  
- Regularization adds penalty term:  

$$
L = L_{data} + \lambda \|W\|_2^2
$$

- **Benefits:**  
  - Prefers simpler models  
  - Prevents overfitting  
  - Improves generalization  

---

## Q10. Comparison Question (10 marks)

**Q:** Compare SVM loss and cross-entropy loss.

**A:**  
- **SVM loss:**  
  - Uses **margins**, focuses on separation.  
  - Loss = 0 if margin is satisfied.  
- **Cross-entropy loss:**  
  - Produces probabilities, encourages confident predictions.  
  - Always > 0 unless probability of true class = 1.  

**Example:** Scores $[10, -2, 3], y=0$ →  
- SVM loss = 0  
- Cross-entropy > 0  

**Conclusion:** Cross-entropy is smoother and works better with gradient descent.

Custom KNN (CPU)

Q: Implement KNN with Euclidean distance. What’s the complexity?

```python
import numpy as np
from scipy.spatial import distance

class CustomKNNCPU:
    def __init__(self, k=3):
        self.k = k   # 🔑 number of neighbors (k=3 → majority vote from 3 nearest)

    def fit(self, X_train, y_train):
        self.X_train = X_train   # 🔑 shape (N,d), e.g. (140,2)
        self.y_train = y_train   # 🔑 shape (N,), e.g. (140,)

    def predict(self, X_test):
        preds = []
        for x in X_test:  # 🔑 x shape = (d,)
            dists = [distance.euclidean(x, xi) for xi in self.X_train]
            k_idx = np.argsort(dists)[:self.k]   # pick k nearest indices
            k_labels = self.y_train[k_idx]       # labels of nearest neighbors
            preds.append(np.bincount(k_labels).argmax())  # majority vote
        return np.array(preds)
```

Custom KNN (GPU with PyTorch)

Q: Why chunk test data for GPU KNN?

```python
import torch

class CustomKNNGPU:
    def __init__(self, k=3, num_classes=10, chunk_size=256):
        self.k = k                 # 🔑 neighbors
        self.num_classes = num_classes  # 🔑 classes (for CIFAR-10 → 10)
        self.chunk_size = chunk_size    # 🔑 batch size for GPU (256 avoids OOM)

    def fit(self, X_train, y_train):
        self.X_train = X_train.to(device)   # 🔑 [N,d], e.g. (1000,3072)
        self.y_train = y_train.long().to(device)

    @torch.no_grad()
    def predict(self, X_test):
        preds = []
        N = X_test.size(0)  # 🔑 e.g. 400
        for start in range(0, N, self.chunk_size):
            end = min(start+self.chunk_size, N)
            xb = X_test[start:end]    # 🔑 shape [chunk_size,d]
            dists = torch.cdist(xb, self.X_train, p=2)  # 🔑 [chunk_size, N]
            _, idx = torch.topk(dists, self.k, largest=False)
            neigh_labels = self.y_train[idx]   # 🔑 [chunk_size, k]
            batch_pred = []
            for row in neigh_labels:
                counts = torch.bincount(row, minlength=self.num_classes)
                batch_pred.append(torch.argmax(counts))
            preds.append(torch.stack(batch_pred))
        return torch.cat(preds)

# Device management
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

## Q5. Perceptron Update Rule

**Q:** Write the update rule for perceptron.

---

### 📘 Theory

The perceptron is a **linear classifier** that updates its parameters only when a sample is **misclassified**.  
Given:
- Weight vector $w$
- Bias $b$
- Learning rate $\eta$
- Training sample $(x_i, y_i)$ where $y_i \in \{-1, +1\}$

The update rule is:

$$
w \leftarrow w + \eta \, y_i \, x_i
$$

$$
b \leftarrow b + \eta \, y_i
$$

---

### 🔑 Notes
- Update happens **only if** $y_i (w \cdot x_i + b) \leq 0$ (i.e., prediction is wrong).  
- $\eta$ controls step size (learning rate).  
- Guarantees convergence for **linearly separable data**.  


```python
class PerceptronCPU:
    def __init__(self, lr=0.01, n_iter=1000):
        self.lr = lr           # 🔑 learning rate (0.01)
        self.n_iter = n_iter   # 🔑 epochs

    def fit(self, X, y):
        n_samples, n_features = X.shape   # 🔑 X shape (N,d)
        self.w = np.zeros(n_features)     # 🔑 weights (d,)
        self.b = 0                        # 🔑 bias scalar
        y_ = np.where(y==0, -1, 1)        # 🔑 convert labels to {-1,1}
        for _ in range(self.n_iter):
            for i, xi in enumerate(X):
                if y_[i]*(np.dot(xi,self.w)+self.b) <= 0:  # misclassified
                    self.w += self.lr*y_[i]*xi
                    self.b += self.lr*y_[i]

    def predict(self, X):
        return np.where(np.dot(X,self.w)+self.b >= 0, 1, 0)
```

Multi-class Perceptron

Q: Extend perceptron to multiple classes.

```python
class MultiPerceptronCPU:
    def __init__(self, lr=0.01, n_iter=100):
        self.lr = lr       # 🔑 learning rate
        self.n_iter = n_iter

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.classes = np.unique(y)     # 🔑 number of classes, e.g. 10
        n_classes = len(self.classes)
        self.W = np.zeros((n_classes, n_features))  # 🔑 shape (C,d)
        self.b = np.zeros(n_classes)                # 🔑 shape (C,)
        for _ in range(self.n_iter):
            for xi, yi in zip(X,y):
                scores = np.dot(self.W,xi)+self.b   # 🔑 shape (C,)
                y_pred = np.argmax(scores)
                if y_pred != yi:
                    self.W[yi] += self.lr*xi
                    self.W[y_pred] -= self.lr*xi

    def predict(self, X):
        scores = np.dot(X,self.W.T)+self.b   # 🔑 [N,C]
        return np.argmax(scores, axis=1)
```
