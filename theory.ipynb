## 🔹 Question 1: Explain Properties of LSIS (Linear Shift-Invariant System)

Let the input be $f(x)$, the output be $g(x)$, and the system be LSIS with impulse response $h(x)$.

### 1. Linearity (Superposition)
- **Additivity:**  
  $$
  \mathcal{L}\{f_1+f_2\}= \mathcal{L}\{f_1\} + \mathcal{L}\{f_2\}
  $$
- **Homogeneity:**  
  $$
  \mathcal{L}\{\alpha f\}= \alpha \,\mathcal{L}\{f\}
  $$
- **Combined:**  
  $$
  \mathcal{L}\Big\{\sum_i \alpha_i f_i\Big\}=\sum_i \alpha_i \mathcal{L}\{f_i\}
  $$

### 2. Shift Invariance
- Shifting input shifts output:  
  $$
  \mathcal{L}\{f(x-x_0)\}=g(x-x_0)
  $$

### 3. Convolution Representation
- 1D system:  
  $$
  g(x)=(f*h)(x)=\int f(\xi)\,h(x-\xi)\,d\xi
  $$
- 2D system:  
  $$
  g(x,y)=\iint f(\xi,\eta)\,h(x-\xi,y-\eta)\,d\xi\,d\eta
  $$

### 4. Impulse Response
- For $\delta(x)$:  
  $$
  \mathcal{L}\{\delta(x)\}=h(x)
  $$

### 5. Frequency Domain Multiplication
- Fourier domain relation:  
  $$
  G(\omega)=H(\omega)\,F(\omega)
  $$
- Complex exponentials are eigenfunctions.

### 6. Algebraic Properties
- Commutative: $f*h=h*f$  
- Associative: $(f*h_1)*h_2=f*(h_1*h_2)$  
- Distributive: $f*(h_1+h_2)=f*h_1+f*h_2$

### 7. Stability
- BIBO stability: if $\int |h|<\infty$, bounded input → bounded output.

### 8. Causality / Locality
- Time signals: causal if output depends only on present/past input.  
- Images: filters are local (finite support), not causal.

### 9. Examples
- **Gaussian blur:**  
  $$
  h(x,y)=\frac{1}{2\pi\sigma^2}\exp\left(-\frac{x^2+y^2}{2\sigma^2}\right)
  $$
- **Derivative filters** (e.g., Sobel).

### 10. Separable Kernels
- If $h(x,y)=h_x(x)\,h_y(y)$, convolution can be done in two 1D steps.

## 🔹 Question 2: Characteristic Scale in SIFT and Blob Detection

### 1. Scale-Space Representation
- Defined as:  
  $$
  L(x,y,\sigma) = G(x,y,\sigma)*I(x,y)
  $$
- Gaussian kernel:  
  $$
  G(x,y,\sigma)=\frac{1}{2\pi\sigma^2}\exp\left(-\frac{x^2+y^2}{2\sigma^2}\right)
  $$

### 2. Blob Detection with Laplacian of Gaussian (LoG)
- Scale-normalized operator:  
  $$
  \mathcal{R}(x,y,\sigma)=\sigma^2\nabla^2 L(x,y,\sigma)
  $$
- DoG approximation:  
  $$
  \text{DoG}(x,y,\sigma)\approx L(x,y,k\sigma)-L(x,y,\sigma)
  $$

### 3. Characteristic Scale ($\sigma^{\ast}$)
- Defined as:  
  $$
  \sigma^{\ast}=\arg\max_{\sigma}\Big|\mathcal{R}(x_0,y_0,\sigma)\Big|
  $$

### 4. Relation Between Blob Size and $\sigma^{\ast}$
- For blob radius $R$:  
  $$
  \sigma^{\ast}\approx\frac{R}{\sqrt{2}}, 
  \quad R \approx \sqrt{2}\,\sigma^{\ast}
  $$

### 5. SIFT Keypoint Detection
- Gaussian pyramid (octaves).  
- DoG across scales.  
- 3D non-maximum suppression.  
- Refine $(x,y,\sigma^{\ast})$.

### 6. Importance of Scale Normalization
- The $\sigma^2$ factor ensures responses are comparable across scales.

### 7. Blob Examples
- Blob A (narrow): maximum at low $\sigma$.  
- Blob B (medium): maximum at mid $\sigma$.  
- Blob C (wide): maximum at high $\sigma$.

### 8. Keypoint Stability
- $\sigma^{\ast}$ gives consistent detection across zoom levels → scale invariance.

### 9. Orientation Assignment
- At $(x^{\ast},y^{\ast},\sigma^{\ast})$, gradients define keypoint orientation → rotation invariance.

### 10. Summary
- **Characteristic scale** = $\sigma^{\ast}$ where normalized LoG/DoG response is extremal.  
- Maps blob size $R\approx \sqrt{2}\,\sigma^{\ast}$.  
- Enables SIFT to detect blobs of different sizes and remain scale invariant.

## 1) Explain properties of LSIS (Linear Shift-Invariant Systems)

Let $f(x)$ be the input image (or 1-D signal), $g(x)$ the output, and $h(x)$ the system’s impulse response (point-spread function, PSF). An LSIS is fully characterized by:

### Linearity (Superposition)
- **Additivity:**
  $$
  \mathcal{L}\{f_1+f_2\} = \mathcal{L}\{f_1\} + \mathcal{L}\{f_2\}
  $$
- **Homogeneity:**
  $$
  \mathcal{L}\{\alpha f\} = \alpha \,\mathcal{L}\{f\}
  $$
- **Together:**
  $$
  \mathcal{L}\Big\{\sum_i \alpha_i f_i\Big\} = \sum_i \alpha_i \mathcal{L}\{f_i\}
  $$

### Shift (Translation) Invariance
- Shifting the input shifts the output:
  $$
  \mathcal{L}\{f(x-x_0)\} = g(x-x_0)
  $$

### Convolution Representation
- 1D:
  $$
  g(x) = (f*h)(x) = \int f(\xi)\,h(x-\xi)\,d\xi
  $$
- 2D:
  $$
  g(x,y) = (f*h)(x,y) = \iint f(\xi,\eta)\,h(x-\xi,y-\eta)\,d\xi\,d\eta
  $$

### Impulse Response
- For Dirac delta $\delta(x)$:
  $$
  \mathcal{L}\{\delta(x)\} = h(x)
  $$

### Frequency-Domain Multiplication
- With Fourier transforms $F(\omega),H(\omega),G(\omega)$:
  $$
  G(\omega) = H(\omega)\,F(\omega)
  $$
- Complex exponentials $e^{j\omega x}$ are eigenfunctions.

### Convolution Algebra
- Commutative: $f*h = h*f$  
- Associative: $(f*h_1)*h_2 = f*(h_1*h_2)$  
- Distributive: $f*(h_1+h_2) = f*h_1 + f*h_2$

### BIBO Stability
If $h \in L^1$ (i.e. $\int |h| < \infty$), bounded input ⇒ bounded output.

### Causality / Locality
- Time signals: causal if output depends only on present/past inputs.  
- Images: often local but not causal.

### Examples
- **Gaussian blur:**
  $$
  h(x,y) = \frac{1}{2\pi\sigma^2}\exp\left(-\frac{x^2+y^2}{2\sigma^2}\right)
  $$
- **Derivative filters** (Sobel/Prewitt).

### Separable Kernels
If $h(x,y) = h_x(x) h_y(y)$ (e.g., Gaussian):
$$
f*h = (f*h_x)*h_y
$$

## 2) What is the characteristic scale, and how does it help detect different blob sizes in SIFT?

Goal: detect the same structure (e.g., a blob) despite zoom (scale) changes.  
SIFT builds a scale space and finds the **characteristic scale** $\sigma^{\ast}$ at which the response is extremal.

### Gaussian Scale Space
- Definition:
  $$
  L(x,y,\sigma) = G(x,y,\sigma)*I(x,y)
  $$
- Gaussian kernel:
  $$
  G(x,y,\sigma)=\frac{1}{2\pi\sigma^2}\exp\left(-\frac{x^2+y^2}{2\sigma^2}\right)
  $$

### Scale-Normalized Blob Operator
- LoG:
  $$
  R(x,y,\sigma) = \sigma^2 \nabla^2 L(x,y,\sigma)
  $$
- DoG approximation:
  $$
  \text{DoG}(x,y,\sigma)\approx L(x,y,k\sigma)-L(x,y,\sigma)\,\propto\,\sigma^2 \nabla^2 L(x,y,\sigma)
  $$

### Characteristic Scale $\sigma^{\ast}$
- Defined at point $(x_0,y_0)$:
  $$
  \sigma^{\ast} = \arg\max_{\sigma}\,\Big|R(x_0,y_0,\sigma)\Big|
  $$

### Relation Between Blob Size and $\sigma^{\ast}$
- For circular blob radius $R$:
  $$
  \sigma^{\ast} \approx \frac{R}{\sqrt{2}}, \quad R \approx \sqrt{2}\,\sigma^{\ast}
  $$

### How SIFT Detects Blobs
- Gaussian pyramid (octaves).  
- DoG across scales.  
- Non-max suppression in 3D $(x,y,\sigma)$.  
- Refinement of location and $\sigma^{\ast}$.

### Why Normalization is Essential
- Without $\sigma^2$, large $\sigma$ values dominate.  
- Normalization ensures true scale-stable extrema.

### From Blobs to SIFT Keypoints
- Extrema $(x^{\ast},y^{\ast},\sigma^{\ast})$ = keypoint.  
- Orientation from local gradients → **rotation + scale invariance**.

### Key Equations (Summary)
- Scale space: $L=G_\sigma * I$  
- DoG: $\text{DoG}(x,y,\sigma)=L(x,y,k\sigma)-L(x,y,\sigma)$  
- Normalized LoG: $\sigma^2\nabla^2 L$  
- Characteristic scale: $\sigma^{\ast}=\arg\max_{\sigma}|\sigma^2\nabla^2 L|$  
- Blob size relation: $R \approx \sqrt{2}\,\sigma^{\ast}$

# 📊 Analysis of Results

---

## 1. Gaussian Filter
- Applies **weighted averaging** with a Gaussian kernel.  
- Smooths both textures and edges uniformly.  
- High $\sigma$ (e.g., $\sigma=50$) → strong blur: fine details and edges are equally smoothed.  
- Edges become blurred and less sharp.

**Mathematical Form:**
$$
G(x,y) = \frac{1}{2\pi\sigma^2} \exp\!\left(-\frac{x^2+y^2}{2\sigma^2}\right)
$$

and

$$
I_{out}(x,y) = \sum_{u,v} I(x-u,\,y-v)\,G(u,v)
$$

---

## 2. Bilateral Filter
- Combines:
  - **Spatial Gaussian weight** (distance between pixels).  
  - **Range Gaussian weight** (intensity similarity).  
- Nearby pixels with **similar intensity** are strongly averaged.  
- Pixels across edges (large intensity difference) get low weight.  
- This preserves **edges** while smoothing noise inside regions.

**Mathematical Form:**
$$
I_{out}(x) \;=\; \frac{1}{W_p} \sum_{x_i \in \Omega} I(x_i)\,
\exp\!\left(-\frac{\|x - x_i\|^2}{2\sigma_s^2}\right)\,
\exp\!\left(-\frac{|I(x)-I(x_i)|^2}{2\sigma_r^2}\right)
$$

where:
- $\sigma_s$ = spatial sigma  
- $\sigma_r$ = range (color/intensity) sigma  
- $W_p$ = normalization factor

---

## 3. Why Bilateral Filter Preserves Edges
- **Gaussian blur:** only spatial distance matters → edges are blurred like noise.  
- **Bilateral filter:** considers both spatial distance and intensity difference.  
  - Neighbors **across edges** → large intensity difference → low weight → edge preserved.  
  - Neighbors **within same region** → small intensity difference → high weight → region smoothed.  

➡️ Result: bilateral filtering is **edge-aware smoothing**, superior for tasks like denoising, cartooning, or preprocessing before edge detection.

---

## 📝 Final Exam-Style Answer
- **Gaussian filter** → smooths entire image, reduces noise but blurs edges.  
- **Bilateral filter** → smooths within regions while preserving sharp edges, due to intensity weighting.  
- Bilateral filtering is **computationally more expensive**, but **superior when edge preservation is critical**.
